# tinycpu framebuffer demo

Load the framebuffer demo through the AXI-Lite loader and display the 10x20 framebuffer mirror.

```python
from pynq import Overlay, MMIO
from tinycpu_loader import TinyCPULoader, find_loader_ip

overlay = Overlay("tinycpu.bit")
overlay.ip_dict
```

```python
# Set preferred_name if auto-discovery does not choose the loader/control IP.
ip = find_loader_ip(overlay, preferred_name=None)
mmio = MMIO(ip["phys_addr"], ip["addr_range"])
cpu = TinyCPULoader(mmio)
```

```python
cpu.reset()
cpu.load_hex("../programs/framebuffer_demo/firmware.hex")
cpu.set_boot_pc(0)
cpu.run()
status, code = cpu.wait_for_test_status(timeout_s=2.0)
print(f"test_status={status:#x}, test_code={code:#x}")
```

```python
from IPython.display import clear_output
import time

for _ in range(100):
    cells = cpu.read_framebuffer()
    clear_output(wait=True)
    cpu.print_board(cells)
    print("game_status:", hex(cpu.read_game_status()))
    print("frame:", cpu.read_frame_counter())
    time.sleep(0.1)
```
