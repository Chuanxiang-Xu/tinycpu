# tinycpu interactive I/O demo

Load a generic interactive program, write HOST_INPUT through AXI-Lite, and render the framebuffer mirror.

```python
from pynq import Overlay, MMIO
from tinycpu_loader import TinyCPULoader, find_loader_ip

overlay = Overlay("tinycpu.bit")
overlay.ip_dict
```

```python
LOADER_IP_NAME = None
ip = find_loader_ip(overlay, preferred_name=LOADER_IP_NAME)
cpu = TinyCPULoader(MMIO(ip["phys_addr"], ip["addr_range"]))

cpu.reset()
cpu.load_hex("../programs/interactive_demo/firmware.hex")
cpu.set_boot_pc(0)
cpu.run()
```

```python
from IPython.display import clear_output
import time

for value in [1, 2, 4, 8, 16, 32, 64, 128]:
    cpu.write_input(value)
    time.sleep(0.05)
    clear_output(wait=True)
    cpu.print_board(cpu.read_framebuffer())
    print("status:", hex(cpu.read_app_status()))
    print("last input:", hex(cpu.read_app_value0()))
    print("frame:", cpu.read_frame_counter())
    time.sleep(0.2)
```
